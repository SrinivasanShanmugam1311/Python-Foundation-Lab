### 1. Print 1 to 5 using recursion


````python
def print_1_to_5(number):
    # Stop after printing 5.
    if number > 5:
        return

    # Print before recursion, so numbers appear while going deeper.
    print(number)
    print_1_to_5(number + 1)

print_1_to_5(1)
````

### 2. Print 5 to 1 using recursion


````python
def print_5_to_1(number):
    # Stop after reaching 0.
    if number == 0:
        return

    print(number)
    print_5_to_1(number - 1)

print_5_to_1(5)
````

### 3. Print 1 to n using recursion


````python
def print_1_to_n(number, target):
    # Stop when the current number goes beyond the target.
    if number > target:
        return

    print(number)
    print_1_to_n(number + 1, target)

print_1_to_n(1, 5)
````

### 4. Print n to 1 using recursion


````python
def print_n_to_1(number):
    # Stop after reaching 0.
    if number == 0:
        return

    print(number)
    print_n_to_1(number - 1)

print_n_to_1(5)
````

### 5. Print even numbers from 2 to 20


````python
def print_even_numbers(number, target):
    # Stop after crossing the target.
    if number > target:
        return

    print(number)
    print_even_numbers(number + 2, target)

print_even_numbers(2, 20)
````

### 6. Print odd numbers from initial value 2 to target 20


````python
def print_odd_numbers(number, target):
    # Stop after crossing the target.
    if number > target:
        return

    # Print only odd values.
    if number % 2 != 0:
        print(number)

    print_odd_numbers(number + 1, target)

print_odd_numbers(2, 20)
````

### 7. Find the sum of 1 to n


````python
def sum_1_to_n(number):
    # Nothing is left to add at 0.
    if number == 0:
        return 0

    return number + sum_1_to_n(number - 1)

print(sum_1_to_n(5))
````

### 8. Find factorial of n


````python
def factorial(number):
    # 0! and 1! are both 1.
    if number <= 1:
        return 1

    return number * factorial(number - 1)

print(factorial(5))
````

### 9. Count the number of digits


````python
def count_digits(number):
    number = abs(number)

    # A single-digit number has one digit.
    if number < 10:
        return 1

    return 1 + count_digits(number // 10)

print(count_digits(12345))
````

### 10. Find the sum of digits


````python
def sum_digits(number):
    number = abs(number)

    # No digits are left at 0.
    if number == 0:
        return 0

    last_digit = number % 10
    remaining_number = number // 10
    return last_digit + sum_digits(remaining_number)

print(sum_digits(1234))
````

### 11. Calculate a raised to power n


````python
def power(base, exponent):
    # Any number raised to 0 is 1.
    if exponent == 0:
        return 1

    return base * power(base, exponent - 1)

print(power(2, 5))
````

### 12. Reverse a string


````python
def reverse_string(text):
    # Empty or one-character text is already reversed.
    if len(text) <= 1:
        return text

    return reverse_string(text[1:]) + text[0]

print(reverse_string("hello"))
````

### 13. Check whether a string is a palindrome


````python
def is_palindrome(text):
    # Zero or one character always reads the same both ways.
    if len(text) <= 1:
        return True

    # First and last characters must match.
    if text[0] != text[-1]:
        return False

    return is_palindrome(text[1:-1])

print(is_palindrome("madam"))
````

### 14. Find the maximum element in a list


````python
def find_maximum(numbers, index=0):
    # Last element becomes the starting maximum while returning.
    if index == len(numbers) - 1:
        return numbers[index]

    maximum_in_rest = find_maximum(numbers, index + 1)
    return max(numbers[index], maximum_in_rest)

print(find_maximum([4, 9, 2, 7]))
````

### 15. Find the sum of elements in a list


````python
def sum_list(numbers, index=0):
    # No elements are left when index reaches the list length.
    if index == len(numbers):
        return 0

    return numbers[index] + sum_list(numbers, index + 1)

print(sum_list([10, 20, 30]))
````

### 16. Search for an element in a list


````python
def search_element(numbers, target, index=0):
    # Target was not found after checking the whole list.
    if index == len(numbers):
        return False

    if numbers[index] == target:
        return True

    return search_element(numbers, target, index + 1)

print(search_element([10, 20, 30, 40], 30))
````

### 17. Generate the Fibonacci series


````python
def fibonacci_series(count, first=0, second=1):
    # Stop after producing the requested number of values.
    if count == 0:
        return

    print(first, end=" ")
    fibonacci_series(count - 1, second, first + second)

fibonacci_series(10)
print()
````

### 18. Perform binary search using recursion


````python
def binary_search(numbers, target, low, high):
    # Search range became empty.
    if low > high:
        return -1

    middle = (low + high) // 2

    if numbers[middle] == target:
        return middle

    if target < numbers[middle]:
        return binary_search(numbers, target, low, middle - 1)

    return binary_search(numbers, target, middle + 1, high)

numbers = [10, 20, 30, 40, 50]
print(binary_search(numbers, 40, 0, len(numbers) - 1))
````

### 19. Find the minimum element in a list


````python
def find_minimum(numbers, index=0):
    if index == len(numbers) - 1:
        return numbers[index]

    minimum_in_rest = find_minimum(numbers, index + 1)
    return min(numbers[index], minimum_in_rest)

print(find_minimum([4, 9, 2, 7]))
````

### 20. Count even numbers in a list


````python
def count_even_numbers(numbers, index=0):
    if index == len(numbers):
        return 0

    current_count = 1 if numbers[index] % 2 == 0 else 0
    return current_count + count_even_numbers(numbers, index + 1)

print(count_even_numbers([1, 2, 4, 5, 6]))
````

### 21. Count odd numbers in a list


````python
def count_odd_numbers(numbers, index=0):
    if index == len(numbers):
        return 0

    current_count = 1 if numbers[index] % 2 != 0 else 0
    return current_count + count_odd_numbers(numbers, index + 1)

print(count_odd_numbers([1, 2, 3, 4, 5]))
````

### 22. Find the product of all elements in a list


````python
def product_list(numbers, index=0):
    # Multiplication identity is 1 when nothing is left.
    if index == len(numbers):
        return 1

    return numbers[index] * product_list(numbers, index + 1)

print(product_list([2, 3, 4]))
````

### 23. Find the sum of digits of a number


````python
def sum_digits_list_section(number):
    number = abs(number)
    if number == 0:
        return 0

    return number % 10 + sum_digits_list_section(number // 10)

print(sum_digits_list_section(5678))
````

### 24. Find the product of digits of a number


````python
def product_digits(number):
    number = abs(number)

    # Last remaining digit is returned directly.
    if number < 10:
        return number

    return (number % 10) * product_digits(number // 10)

print(product_digits(234))
````

### 25. Count occurrences of an element in a list


````python
def count_occurrences(numbers, target, index=0):
    if index == len(numbers):
        return 0

    current_count = 1 if numbers[index] == target else 0
    return current_count + count_occurrences(numbers, target, index + 1)

print(count_occurrences([2, 3, 2, 4, 2], 2))
````

### 26. Find the first occurrence of an element


````python
def first_occurrence(numbers, target, index=0):
    if index == len(numbers):
        return -1

    if numbers[index] == target:
        return index

    return first_occurrence(numbers, target, index + 1)

print(first_occurrence([5, 8, 3, 8], 8))
````

### 27. Find the last occurrence of an element


````python
def last_occurrence(numbers, target, index=None):
    if index is None:
        index = len(numbers) - 1

    if index < 0:
        return -1

    if numbers[index] == target:
        return index

    return last_occurrence(numbers, target, index - 1)

print(last_occurrence([5, 8, 3, 8], 8))
````

### 28. Check whether a list is sorted


````python
def is_sorted(numbers, index=0):
    # Reaching the last element means every pair was in order.
    if index >= len(numbers) - 1:
        return True

    if numbers[index] > numbers[index + 1]:
        return False

    return is_sorted(numbers, index + 1)

print(is_sorted([1, 2, 3, 5, 8]))
````

### 29. Remove all occurrences of an element from a list


````python
def remove_occurrences(numbers, target, index=0):
    if index == len(numbers):
        return []

    remaining_result = remove_occurrences(numbers, target, index + 1)

    if numbers[index] == target:
        return remaining_result

    return [numbers[index]] + remaining_result

print(remove_occurrences([1, 2, 3, 2, 4, 2], 2))
````

### 30. Find the second largest element in a list


````python
def collect_unique_values(numbers, index=0):
    if index == len(numbers):
        return []

    remaining_values = collect_unique_values(numbers, index + 1)

    if numbers[index] in remaining_values:
        return remaining_values

    return [numbers[index]] + remaining_values

numbers = [10, 40, 20, 40, 30]
unique_values = collect_unique_values(numbers)
unique_values.sort(reverse=True)
second_largest = unique_values[1] if len(unique_values) >= 2 else None
print(second_largest)
````

### 31. Find the second smallest element in a list


````python
def collect_unique_values_for_smallest(numbers, index=0):
    if index == len(numbers):
        return []

    remaining_values = collect_unique_values_for_smallest(numbers, index + 1)

    if numbers[index] in remaining_values:
        return remaining_values

    return [numbers[index]] + remaining_values

numbers = [10, 40, 20, 10, 30]
unique_values = collect_unique_values_for_smallest(numbers)
unique_values.sort()
second_smallest = unique_values[1] if len(unique_values) >= 2 else None
print(second_smallest)
````

### 33. Check whether two lists are equal


````python
def lists_are_equal(first, second, index=0):
    if len(first) != len(second):
        return False

    if index == len(first):
        return True

    if first[index] != second[index]:
        return False

    return lists_are_equal(first, second, index + 1)

print(lists_are_equal([1, 2, 3], [1, 2, 3]))
````

### 38. Replace a character in a string


````python
def replace_character(text, old_character, new_character):
    if text == "":
        return ""

    current_character = new_character if text[0] == old_character else text[0]
    return current_character + replace_character(text[1:], old_character, new_character)

print(replace_character("banana", "a", "o"))
````

### 39. Convert decimal to binary


````python
def decimal_to_binary(number):
    if number < 2:
        return str(number)

    return decimal_to_binary(number // 2) + str(number % 2)

print(decimal_to_binary(13))
````

### 40. Convert binary to decimal


````python
def binary_to_decimal(binary_text, index=0):
    if index == len(binary_text):
        return 0

    current_digit = int(binary_text[index])
    power = len(binary_text) - index - 1

    return current_digit * (2 ** power) + binary_to_decimal(binary_text, index + 1)

print(binary_to_decimal("1101"))
````

### 41. Find GCD of two numbers


````python
def gcd(first, second):
    # When second becomes 0, first is the GCD.
    if second == 0:
        return abs(first)

    return gcd(second, first % second)

print(gcd(48, 18))
````

### 42. Find LCM of two numbers


````python
def gcd_for_lcm(first, second):
    if second == 0:
        return abs(first)

    return gcd_for_lcm(second, first % second)


def lcm(first, second):
    if first == 0 or second == 0:
        return 0

    return abs(first * second) // gcd_for_lcm(first, second)

print(lcm(12, 18))
````

### 43. Perform binary search using recursion


````python
def recursive_binary_search(numbers, target, low, high):
    if low > high:
        return -1

    middle = (low + high) // 2

    if numbers[middle] == target:
        return middle

    if target < numbers[middle]:
        return recursive_binary_search(numbers, target, low, middle - 1)

    return recursive_binary_search(numbers, target, middle + 1, high)

numbers = [2, 4, 6, 8, 10, 12]
print(recursive_binary_search(numbers, 8, 0, len(numbers) - 1))
````

### 45. Find the nth Fibonacci number


````python
def nth_fibonacci(position):
    # Using 0-based Fibonacci positions: F(0)=0, F(1)=1.
    if position <= 1:
        return position

    return nth_fibonacci(position - 1) + nth_fibonacci(position - 2)

print(nth_fibonacci(7))
````

### 46. Find the sum of elements in a nested list


````python
def sum_nested_list(items):
    if items == []:
        return 0

    first_item = items[0]
    remaining_items = items[1:]

    if isinstance(first_item, list):
        return sum_nested_list(first_item) + sum_nested_list(remaining_items)

    return first_item + sum_nested_list(remaining_items)

print(sum_nested_list([1, [2, 3], [4, [5]]]))
````

### 47. Flatten a nested list


````python
def flatten_nested_list(items):
    if items == []:
        return []

    first_item = items[0]
    remaining_items = items[1:]

    if isinstance(first_item, list):
        return flatten_nested_list(first_item) + flatten_nested_list(remaining_items)

    return [first_item] + flatten_nested_list(remaining_items)

print(flatten_nested_list([1, [2, 3], [4, [5]]]))
````

### 48. Count all elements in a nested list


````python
def count_nested_elements(items):
    if items == []:
        return 0

    first_item = items[0]
    remaining_items = items[1:]

    if isinstance(first_item, list):
        return count_nested_elements(first_item) + count_nested_elements(remaining_items)

    return 1 + count_nested_elements(remaining_items)

print(count_nested_elements([1, [2, 3], [4, [5]]]))
````

### 49. Generate all permutations of a string


````python
def string_permutations(text):
    # One character has only one permutation.
    if len(text) <= 1:
        return [text]

    result = []

    for index in range(len(text)):
        current_character = text[index]
        remaining_text = text[:index] + text[index + 1:]

        for smaller_permutation in string_permutations(remaining_text):
            result.append(current_character + smaller_permutation)

    return result

print(string_permutations("ABC"))
````

### 50. Generate combinations of a given size from a string


````python
def string_combinations(text, size):
    # A combination of size 0 is the empty choice.
    if size == 0:
        return [""]

    # Not enough characters remain to complete the combination.
    if len(text) < size:
        return []

    # CHOOSE the first character.
    with_first = [
        text[0] + combination
        for combination in string_combinations(text[1:], size - 1)
    ]

    # SKIP the first character.
    without_first = string_combinations(text[1:], size)

    return with_first + without_first

print(string_combinations("ABCD", 2))
````

````python
def decorator(function):

    def wrapper():
        print("Before function")

        function()

        print("After function")

    return wrapper


def greet():
    print("Hello")


greet = decorator(greet)

greet()
````

````python
def decorator(function):

    def wrapper(name):
        print("Before function")

        function(name)

        print("After function")

    return wrapper


@decorator
def greet(name):
    print("Hello", name)


greet("Alex")
````

````python
def decorator(function):

    def wrapper(*args):
        print("Before function")

        function(*args)

        print("After function")

    return wrapper


@decorator
def add(a, b):
    print(a + b)


add(10, 20)
````

````python
def decorator(function):

    def wrapper(*args, **kwargs):
        print("Before function")

        result = function(*args, **kwargs)

        print("After function")

        return result

    return wrapper


@decorator
def greet(name, age):
    print(name, age)


greet("Alex", 45)
````
